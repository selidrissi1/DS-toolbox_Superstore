```python
#import numpy
import numpy as np
import random
import time
```

```python
S = [252, 252, 252, 252, 228, 228, 228, 180, 180, 180, 140, 140, 140, 120, 120, 120, 120] # S (list of items)
V = [600, 600, 600, 600, 600, 600] # V → list of available bins
T = [0, 0, 0, 0, 0, 0] # T → list of bins used
```

### **First-Fit Algorithm**

## Random Order (S)

```python
shuffled_S = S.copy()
random.shuffle(shuffled_S)
```

```python
print(shuffled_S)
```

```python
def first_fit(S, V, bin_capacity):
    T = []  # List of bins already used
    while S:
        item = S.pop(0)  # Extract an item from the top of list S
        # Flag to check if the item has been packed
        packed = False
        for bin_index, bin_capacity_left in enumerate(T):
            # If the item can fit in the current bin
            if item <= bin_capacity_left:
                T[bin_index] -= item  # Reduce the bin capacity
                packed = True
                break
        if not packed:
            # If the item cannot fit in any existing bin, create a new bin
            new_bin_capacity = bin_capacity - item
            T.append(new_bin_capacity)
        if not S:  # If S is empty, all items have been loaded
            return T, V
    return T, V

# Example usage
V = []
bin_capacity = 600

# Start timer
start_time = time.time()

packed_bins, unused_bins = first_fit(shuffled_S, V, bin_capacity)

# End timer
end_time = time.time()


print("Bins used:")
for i, bin_capacity_left in enumerate(packed_bins):
    print(f"Bin {i+1}: Remaining capacity {bin_capacity_left}")
print("Unused bins:")
for i, bin_capacity in enumerate(unused_bins):
    print(f"Bin {i+1}: Capacity {bin_capacity}")

# Calculate elapsed time
elapsed_time = end_time - start_time
print("Elapsed time (in seconds): ", elapsed_time)
```
